## DATA VISULIZATION

###### 1.SALES BY CATEGORY

##### Top 10 Products by Sales

```python
product_sales = df.groupby("Product Name")["Sales"].sum()
```

```python
top10_sales = product_sales.sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10,6))

plt.barh(top10_sales.index, top10_sales.values)

plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.title("Top 10 Products by Sales")

plt.gca().invert_yaxis()

plt.show()
```

##### Top 10 Products by Profit

```python
product_profit = df.groupby("Product Name")["Profit"].sum()
```

```python
top10_profit = product_profit.sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10,6))

plt.barh(top10_profit.index, top10_profit.values)

plt.xlabel("Total Profit")
plt.ylabel("Product")
plt.title("Top 10 Products by Profit")

plt.gca().invert_yaxis()

plt.show()
```
